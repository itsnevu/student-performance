```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import os

# 1. LOAD DATA MAHASISWA
# Menggunakan sep=',' karena file student_dropout_prediction.csv menggunakan koma, bukan titik koma.
file_path = '../data/raw/student_dropout_prediction.csv'
df = pd.read_csv(file_path, sep=',')

# 2. STATISTIK DESKRIPTIF (SAINS)
print("--- Deskripsi Statistik Mahasiswa ---")
display(df.describe())

# 3. ANALISIS TARGET (Penting untuk Audit Ketimpangan)
# Sekarang kolom 'Target' sudah terbaca dengan benar.
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='Target', palette='viridis')
plt.title('Proporsi Kelulusan vs Dropout Mahasiswa')
plt.show()

# 4. MATRIKS KORELASI (SAINS: Mengapa mereka Dropout?)
plt.figure(figsize=(15, 10))
# Mengambil kolom numerik untuk melihat hubungan IPK, Kredit, dll.
corr = df.select_dtypes(include=['number']).corr()
sns.heatmap(corr, annot=False, cmap='RdBu', center=0)
plt.title('Analisis Korelasi Fitur Akademik Mahasiswa')
plt.show()

# 5. ANALISIS VARIABEL KUNCI (Contoh: Pengaruh Nilai Semester 1)
plt.figure(figsize=(10, 6))
# Nama kolom harus sesuai dengan dataset (case-sensitive)
sns.boxplot(data=df, x='Target', y='Curricular units 1st sem (grade)', palette='Set2')
plt.title('Pengaruh Nilai Semester 1 terhadap Status Kelulusan')
plt.show()
```
